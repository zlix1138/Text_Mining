```python
import pandas as pd
```

```python
pd.set_option('display.max_rows', None)
#임금 minmax 정규화

df = pd.read_csv("quality_jobs_clean.csv")

# 1. 연봉 중간값 계산
df["salary_mid"] = (df["salary_annual_min"] + df["salary_annual_max"]) / 2

# 2. 지역 평균으로 결측치 대체
region_mean = df.groupby("region")["salary_mid"].transform("mean")

df["salary_mid"] = df["salary_mid"].fillna(region_mean)

# 3. 그래도 비어있으면 전체 평균으로 대체(지역의 평균이 비어있을경우)
df["salary_mid"] = df["salary_mid"].fillna(df["salary_mid"].mean())

# 4. Min-Max 정규화
min_salary = df["salary_mid"].min()
max_salary = df["salary_mid"].max()

df["wage_score"] = (df["salary_mid"] - min_salary) / (max_salary - min_salary)


#연 minmax에 가중치 곱하기
df["wage_score_weighted"] = df["wage_score"] * 0.25

print(df[[
    "region",
    "salary_annual_min",
    "salary_annual_max",
    "salary_mid",
    "wage_score",
    "wage_score_weighted"
]])
```

```python
#각 공고의 대
sample = (
    df.groupby("region")
      .first()
      .reset_index()
)

print(sample[[
    "region",
    "salary_annual_min",
    "salary_annual_max",
    "salary_mid",
    "wage_score",
    "wage_score_weighted"
]])
```

```python
#안정성 
df["stability_score"] = (0.5 * df["employment_stability"]+ 0.5 * df["is_public_sector"])
#가중치 곱
df["stability_weighted"] = (df["stability_score"] * 0.25)
pd.set_option('display.width', 1000)

print(df[["region","stability_score","employment_stability","is_public_sector","stability_weighted"]])
```

```python
#각 공고별 대표
sample = (
    df.groupby("region")
      .first()
      .reset_index()
)

print(sample[[
    "region",
    "employment_stability",
    "is_public_sector",
    "stability_score",
    "stability_weighted"
]])
```

```python
#성장성
min_growth = df["q_growth"].min()
max_growth = df["q_growth"].max()
#minmax 정규화
df["growth_score"] = (df["q_growth"] - min_growth) / (max_growth - min_growth)
#가중치
df["growth_weighted"] = (df["growth_score"] * 0.15)

print(df[["region","q_growth","growth_score","growth_weighted"]])
```

```python
sample = (
    df.groupby("region")
      .first()
      .reset_index()
)

print(sample[[
    "region",
    "q_growth",
    "growth_score",
    "growth_weighted"
]])
```

```python
# 복지
min_welfare = df["q_welfare"].min()
max_welfare = df["q_welfare"].max()
df["welfare_score"] = (df["q_welfare"] - min_welfare) / (max_welfare - min_welfare)

# 워라밸
min_welfare = df["q_work_time"].min()
max_welfare = df["q_work_time"].max()
df["work_time_score"] = (df["q_work_time"] - min_welfare) / (max_welfare - min_welfare)

# 복지·워라밸 통합
df["worklife_score"] = (0.5 * df["welfare_score"]+ 0.5 * df["work_time_score"]) 
#가중치
df["worklife_weighted"] = df["worklife_score"] * 0.25
print(df[["region","q_welfare","q_work_time","welfare_score","work_time_score","worklife_score","worklife_weighted"]])
```

```python
sample = (
    df.groupby("region")
      .first()
      .reset_index()
)

print(sample[[
    "region","q_welfare","q_work_time","welfare_score","work_time_score","worklife_score","worklife_weighted"
]])
```

```python
#청년친화성
# 청년친화성 점수
min_youth = df["q_youth_friendly"].min()
max_youth = df["q_youth_friendly"].max()

df["youth_keyword_score"] = ((df["q_youth_friendly"] - min_youth)/(max_youth - min_youth))

# 청년친화성 점수
df["youth_score"] = (0.5 * df["youth_keyword_score"]+ 0.5 * df["is_youth_friendly"])

# 최종 가중치 적용 (10%)
df["youth_weighted"] = (df["youth_score"] * 0.10)

df["youth_score"] = (0.5 * df["youth_keyword_score"]+ 0.5 * df["is_youth_friendly"])

print(df[["region","q_youth_friendly","is_youth_friendly","youth_keyword_score","youth_weighted","youth_score"]])
```

```python
sample = (
    df.groupby("region")
      .first()
      .reset_index()
)

print(sample[[
    "region","q_youth_friendly","is_youth_friendly","youth_keyword_score","youth_weighted","youth_score"
]])
```

```python
#지역별 공고매력도 평균
#모두 합치기
df["job_score"] = (
    df["wage_score_weighted"]
    + df["stability_weighted"]
    + df["growth_weighted"]
    + df["worklife_weighted"]
    + df["youth_weighted"]
)
#평균내기
region_quality = (
    df.groupby("region")["job_score"]
    .mean()
    .reset_index(name="RegionQuality")
)

print(region_quality)
```

```python
#일자리양 점수
import numpy as np

quantity_df = pd.read_csv("quantity_region_jobgroup.csv")
min_quantity = quantity_df["jk_region_total"].min()
max_quantity = quantity_df["jk_region_total"].max()
#minmax
quantity_df["QuantityScore"] = ((quantity_df["jk_region_total"] - min_quantity)/(max_quantity - min_quantity))

quantity_score = (
    quantity_df[["region", "jk_region_total", "QuantityScore"]]
    .drop_duplicates(subset=["region"])
    .reset_index(drop=True)
)

print(quantity_score) 
```

```python
#직무 다양성

# Shannon 값
def diversity_score(x):
    p = x / x.sum()      # 각 직무군 비율
    return -np.sum(p * np.log(p)) 

diversity_df = (
    quantity_df
    .groupby("region")["sr_search_total_max"] #지역별로 묶으라는 뜻
    .apply(diversity_score) 
    .reset_index(name="diversity_raw") #열 이름
)
#minmax
min_div = diversity_df["diversity_raw"].min()
max_div = diversity_df["diversity_raw"].max()

diversity_df["DiversityScore"] = ((diversity_df["diversity_raw"] - min_div)/(max_div - min_div))

print(diversity_df)
```

```python
#민감도 측정 해서 가중치 만들
final_df = (
    region_quality
    .merge(
        quantity_score[["region", "QuantityScore"]],
        on="region" #region을 기준으로 합침
    )
    .merge(
        diversity_df[["region", "DiversityScore"]],
        on="region"
    )
)
print(final_df)

#1
final_df["FinalScore_50_30_20"] = (
    0.5 * final_df["RegionQuality"]
    + 0.3 * final_df["QuantityScore"]
    + 0.2 * final_df["DiversityScore"]
)
#2
final_df["FinalScore_60_20_20"] = (
    0.6 * final_df["RegionQuality"]
    + 0.2 * final_df["QuantityScore"]
    + 0.2 * final_df["DiversityScore"]
)
#3
final_df["FinalScore_40_30_30"] = (
    0.4 * final_df["RegionQuality"]
    + 0.3 * final_df["QuantityScore"]
    + 0.3 * final_df["DiversityScore"]
)
#4
final_df["FinalScore_80_10_10"] = (
    0.8 * final_df["RegionQuality"]
    + 0.1 * final_df["QuantityScore"]
    + 0.1 * final_df["DiversityScore"]
)
final_df["FinalScore_10_80_10"] = (
    0.1 * final_df["RegionQuality"]
    + 0.8 * final_df["QuantityScore"]
    + 0.1 * final_df["DiversityScore"]
)
final_df["FinalScore_10_10_80"] = (
    0.1 * final_df["RegionQuality"]
    + 0.1 * final_df["QuantityScore"]
    + 0.8 * final_df["DiversityScore"]
)
print(final_df[[
    "region",
    "FinalScore_50_30_20",
    "FinalScore_60_20_20",
    "FinalScore_40_30_30",
    "FinalScore_80_10_10",
    "FinalScore_10_80_10",
    "FinalScore_10_10_80",
    
]])
```
